## Exercise 3
(10 points) Suppose you are running `AdaBoost` with 4 training examples. At the start of the current iteration, the four examples have the weights shown in the following table. Another column says if the weak classifier got them correct or incorrect. Determine the new weights for these four examples, and fill in the corresponding entries in the table. Show all your work.

| Observation | Old Weight | Correct? | New Weight |
|:-----------:|:----------:|:--------:|:----------:|
| 1 | 0.16 | Correct   |  |
| 2 | 0.64 | Correct   |  |
| 3 | 0.08 | Incorrect |  |
| 4 | 0.12 | Incorrect |  |


**Work / Answer:** *(write your work here; you may use the code cell below)*

```python
# Exercise 3 work (optional)


import numpy as np

old_weight = np.array([.16, .62, .08, .12])
correct = np.array([True, True, False, False])

error = np.sum(old_weight[~correct])
new_weight = old_weight.copy()
new_weight [correct] *= np.sqrt(error/(1-error))
new_weight [~correct] *= np.sqrt((1-error)/error)

new_weight = new_weight/np.sum(new_weight)
print(new_weight)

# New weight would be 0.1, 0.4, 0.2, 0.3
```

**(a)** (5 points) Using the `pandas` library, read the csv data file and create a data-frame called `heart`. Remove the observations with missing values.

```python
import pandas as pd

heart = pd.read_csv("framingham.csv")
heart = heart.dropna().reset_index(drop=True)

heart.head()
```

**(b)** (50 points) Based on the different interactions that we have so far with this data file, it seems that `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` are the most important predictor variables when it comes to predicting the likelihood of `TenYearCHD`. Then do the following:

1. Using `TenYearCHD` as the target variable and `age`, `totChol`, `sysBP`, `BMI`, `heartRate`, and `glucose` as the input variables, create a `StratifiedKFold` cross validation object with 10 folds (shuffled, with a fixed `random_state`) so that every fold keeps the proportion of 0s and 1s of the target.
2. In each fold, train on the training folds and predict the risk of coronary disease for the observations in the held-out fold, for each of the following models:
   - `RandomForestClassifier` (`n_estimators = 500`, `max_depth = 5`)
   - `ExtraTreesClassifier` (`n_estimators = 500`, `max_depth = 5`)
   - `AdaBoostClassifier` (`n_estimators = 500`, `learning_rate = 0.01`, with decision trees of `max_depth = 3` as weak learners)
   - `GradientBoostingClassifier` (`n_estimators = 500`, `max_depth = 3`, `learning_rate = 0.01`)
3. Using 10% as the cutoff value (i.e., predict 1 when the estimated probability of `TenYearCHD = 1` is at least 0.10), compute the accuracy and recall on the held-out fold.
4. Average the accuracy and recall over the 10 folds for each model. What model would you use to predict `TenYearCHD` based on the accuracy and recall?

```python
from sklearn.model_selection import StratifiedKFold

X = heart[['age', 'totChol', 'sysBP', 'BMI', 'heartRate', 'glucose']]

y = heart['TenYearCHD']

skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
```

```python
# random forest

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, recall_score

rf_md = RandomForestClassifier(n_estimators=500,
                               max_depth=5,
                               random_state=42,
                               n_jobs=-1)
rf_accuracies =[]
rf_recalls = []
for i, (train_index, test_index) in enumerate(skf.split(X, y)):
    
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    rf_md.fit(X_train, y_train)
    y_pred = rf_md.predict_proba(X_test)[:, 1]
    y_pred = (y_pred > 0.1).astype(int)

    accuracy = accuracy_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    rf_accuracies.append(accuracy)
    rf_recalls.append(recall)
    print(f"Fold {i+1} Accuracy: {accuracy}, Recall: {recall}")

print(f"Average Accuracy: {sum(rf_accuracies)/len(rf_accuracies)}")
print(f"Average Recall: {sum(rf_recalls)/len(rf_recalls)}")
```

```python
# extra trees
from sklearn.ensemble import ExtraTreesClassifier
from sklearn.metrics import accuracy_score, recall_score

et_md = ExtraTreesClassifier(n_estimators=500,
                             max_depth=5,
                             random_state=42,
                             n_jobs=-1)

# Running cross-validation 
et_accuracies = []
et_recalls = []
for i, (train_index, test_index) in enumerate(skf.split(X, y)):
    
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    et_md.fit(X_train, y_train)
    y_pred = et_md.predict_proba(X_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    accuracy = accuracy_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    et_accuracies.append(accuracy)
    et_recalls.append(recall)
    print(f"Fold {i+1} Accuracy: {accuracy}, Recall: {recall}")

print(f"Average Accuracy: {sum(et_accuracies)/len(et_accuracies)}")
print(f"Average Recall: {sum(et_recalls)/len(et_recalls)}")
```

```python
# adaboost classifier
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, recall_score

ab_md = AdaBoostClassifier(estimator = DecisionTreeClassifier(max_depth=3, random_state=42),
                           n_estimators=500,
                           random_state=42,
                           learning_rate=0.01)


# Running cross-validation 
ab_accuracies = []
ab_recalls = []
for i, (train_index, test_index) in enumerate(skf.split(X, y)):
    
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    ab_md.fit(X_train, y_train)
    y_pred = ab_md.predict_proba(X_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    accuracy = accuracy_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    ab_accuracies.append(accuracy)
    ab_recalls.append(recall)
    print(f"Fold {i+1} Accuracy: {accuracy}, Recall: {recall}")

print(f"Average Accuracy: {sum(ab_accuracies)/len(ab_accuracies)}")
print(f"Average Recall: {sum(ab_recalls)/len(ab_recalls)}")
```

```python
# gradient boosting classifier

from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import accuracy_score, recall_score

gb_md = GradientBoostingClassifier(n_estimators=500,
                                   max_depth=3,
                                   random_state=42,
                                   learning_rate=0.01)

# Running cross-validation 
gb_accuracies = []
gb_recalls = []
for i, (train_index, test_index) in enumerate(skf.split(X, y)):
    
    X_train, X_test = X.iloc[train_index], X.iloc[test_index]
    y_train, y_test = y.iloc[train_index], y.iloc[test_index]
    
    gb_md.fit(X_train, y_train)
    y_pred = gb_md.predict_proba(X_test)[:, 1]
    y_pred = (y_pred >= 0.1).astype(int)
    
    accuracy = accuracy_score(y_test, y_pred)
    recall = recall_score(y_test, y_pred)
    gb_accuracies.append(accuracy)
    gb_recalls.append(recall)
    print(f"Fold {i+1} Accuracy: {accuracy}, Recall: {recall}")

print(f"Average Accuracy: {sum(gb_accuracies)/len(gb_accuracies)}")
print(f"Average Recall: {sum(gb_recalls)/len(gb_recalls)}")
```
